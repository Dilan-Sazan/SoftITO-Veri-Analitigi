# Pandas Pratik: Dünya Mutluluk Raporu Analizi 🌍

Bu notebook, **Pandas Eğitim** dersinde öğrenilen kavramların gerçek bir veri seti üzerinde uygulamasıdır.

**Veri seti:** Dünya Mutluluk Raporu (2015–2019) — Her yıl için ülkelerin mutluluk skorları ve bunu etkileyen faktörler (GSYH, sosyal destek, sağlık, özgürlük, cömertlik, yolsuzluk algısı).

Derste öğrendiğimiz konuları burada uyguluyoruz:
1. Veri okuma (`read_csv`) ve inceleme (`head`, `info`, `describe`)
2. Series yapısı
3. Veri seçme (`loc`, `iloc`) ve filtreleme
4. Yeni sütun ekleme ve düzenleme
5. Sıralama (`sort_values`) ve gruplama (`groupby`)
6. Birleştirme (`concat` ve `merge`)
7. Uygulamalı analiz: Türkiye'nin 5 yıllık mutluluk trendi

## 1. Veri Okuma ve İnceleme

2019 yılı verisiyle başlıyoruz.

```python
import pandas as pd   # pandas kütüphanesini "pd" takma adıyla içe aktarıyoruz

print("Pandas versiyonu:", pd.__version__)
```

```python
# CSV dosyasını DataFrame olarak okuyoruz
df19 = pd.read_csv("2019.csv")

df19.head()   # ilk 5 satırı gösterir
```

```python
print("Boyut (satır, sütun):", df19.shape)
print()
df19.info()   # sütun isimleri, veri tipleri ve dolu değer sayıları
```

```python
df19.describe()   # sayısal sütunların özet istatistikleri
```

## 2. Series Yapısı

DataFrame'in tek bir sütunu aslında bir **Series**'tir.

```python
skorlar = df19["Score"]     # tek sütun seçince Series elde ederiz

print(type(skorlar))
print("İlk 5 skor:\n", skorlar.head())
print()
print("Ortalama :", round(skorlar.mean(), 3))
print("En yüksek:", skorlar.max())
print("En düşük :", skorlar.min())
```

## 3. Veri Seçme: loc ve iloc

- `iloc` → konuma (pozisyona) göre seçim
- `loc` → etikete/koşula göre seçim

```python
print("--- iloc: ilk satır (dünyanın en mutlu ülkesi) ---")
print(df19.iloc[0])
```

```python
# loc ile koşullu seçim: Türkiye'nin satırını bulalım
turkiye = df19.loc[df19["Country or region"] == "Turkey"]
turkiye
```

```python
# Sadece belirli sütunları seçme
df19[["Country or region", "Score", "GDP per capita"]].head()
```

## 4. Filtreleme

Koşullara uyan satırları seçiyoruz (derste gördüğümüz boolean filtreleme).

```python
# Skoru 7'nin üzerinde olan "çok mutlu" ülkeler
cok_mutlu = df19[df19["Score"] > 7]
print("Skoru 7+ olan ülke sayısı:", len(cok_mutlu))
cok_mutlu[["Overall rank", "Country or region", "Score"]]
```

```python
# Birden fazla koşul: GSYH'si düşük (<1) AMA skoru yüksek (>6) olan ülkeler
ilginc = df19[(df19["GDP per capita"] < 1) & (df19["Score"] > 6)]
ilginc[["Country or region", "Score", "GDP per capita", "Social support"]]
```

## 5. Yeni Sütun Ekleme

Skora göre kategori sütunu oluşturalım.

```python
# pd.cut: sayısal değerleri aralıklara bölerek kategorik hale getirir
df19["Mutluluk_Kategorisi"] = pd.cut(df19["Score"],
                                     bins=[0, 4.5, 6, 10],
                                     labels=["Düşük", "Orta", "Yüksek"])

df19["Mutluluk_Kategorisi"].value_counts()   # her kategoride kaç ülke var
```

## 6. Sıralama ve Gruplama

```python
# En cömert 5 ülke (Generosity sütununa göre azalan sıralama)
df19.sort_values("Generosity", ascending=False)[["Country or region", "Generosity", "Score"]].head()
```

```python
# groupby: kategoriye göre gruplayıp her grubun ortalamalarını alalım
df19.groupby("Mutluluk_Kategorisi", observed=True)[["GDP per capita", "Social support", "Healthy life expectancy"]].mean().round(3)
```

**Yorum:** Mutluluk kategorisi yükseldikçe GSYH, sosyal destek ve sağlık ortalamalarının da arttığı net şekilde görülüyor.

## 7. Merge: Bölge Bilgisini Ekleme

2019 verisinde bölge (Region) sütunu yok, ama 2015 verisinde var. Derste öğrendiğimiz `merge` ile iki tabloyu birleştirip 2019 verisine bölge bilgisi ekleyelim.

```python
df15 = pd.read_csv("2015.csv")   # 2015 verisini okuyoruz (Region sütunu var)

# Ülke ismi üzerinden birleştirme (SQL'deki JOIN mantığı)
birlesik = pd.merge(df19,
                    df15[["Country", "Region"]],          # 2015'ten sadece ülke ve bölge sütunları
                    left_on="Country or region",           # soldaki tabloda eşleşecek sütun
                    right_on="Country",                    # sağdaki tabloda eşleşecek sütun
                    how="left")                            # sol tablodaki tüm satırlar korunur

print("Birleşik tablo boyutu:", birlesik.shape)
birlesik[["Country or region", "Region", "Score"]].head()
```

```python
# Artık bölgelere göre ortalama mutluluk skorunu hesaplayabiliriz
bolge_ort = birlesik.groupby("Region")["Score"].agg(["mean", "count"]).round(3).sort_values("mean", ascending=False)
bolge_ort
```

## 8. Concat: Yılları Alt Alta Birleştirme

2018 ve 2019 dosyalarının sütunları aynı; `concat` ile alt alta ekleyelim.

```python
df18 = pd.read_csv("2018.csv")

# Hangi satırın hangi yıla ait olduğunu bilmek için önce Yil sütunu ekliyoruz
df18["Yil"] = 2018
df19["Yil"] = 2019

iki_yil = pd.concat([df18, df19], ignore_index=True)   # ignore_index: indeksleri 0'dan yeniden numaralandır
print("2018 satır:", len(df18), "| 2019 satır:", len(df19), "| Birleşik:", len(iki_yil))
iki_yil.groupby("Yil")["Score"].mean().round(3)   # yıllara göre dünya ortalaması
```

## 9. Uygulamalı Analiz: Türkiye'nin 5 Yıllık Trendi

Her yılın dosyasında sütun isimleri farklı olduğu için, döngüyle her dosyayı okuyup Türkiye'nin skorunu ve sırasını çekiyoruz.

```python
# Her yıl için (ülke sütunu, skor sütunu, sıra sütunu) isimleri farklı
yil_ayarlari = {
    2015: ("Country",           "Happiness Score", "Happiness Rank"),
    2016: ("Country",           "Happiness Score", "Happiness Rank"),
    2017: ("Country",           "Happiness.Score", "Happiness.Rank"),
    2018: ("Country or region", "Score",           "Overall rank"),
    2019: ("Country or region", "Score",           "Overall rank"),
}

kayitlar = []
for yil, (ulke_sutun, skor_sutun, sira_sutun) in yil_ayarlari.items():
    df = pd.read_csv(f"{yil}.csv")
    tr = df[df[ulke_sutun] == "Turkey"].iloc[0]      # Türkiye'nin satırı
    kayitlar.append({"Yil": yil,
                     "Skor": tr[skor_sutun],
                     "Sira": int(tr[sira_sutun]),
                     "Toplam_Ulke": len(df)})

tr_trend = pd.DataFrame(kayitlar)
tr_trend
```

```python
# En iyi ve en kötü yılları bulalım
print("Türkiye'nin en mutlu yılı :", int(tr_trend.loc[tr_trend['Skor'].idxmax(), 'Yil']),
      "- skor:", tr_trend['Skor'].max())
print("Türkiye'nin en mutsuz yılı:", int(tr_trend.loc[tr_trend['Skor'].idxmin(), 'Yil']),
      "- skor:", tr_trend['Skor'].min())
print("5 yıllık ortalama skor    :", round(tr_trend['Skor'].mean(), 3))
```
