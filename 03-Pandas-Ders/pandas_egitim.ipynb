```python
# Gerekli kütüphaneleri içe aktar
import pandas as pd
import numpy as np

print("Pandas sürümü:", pd.__version__)
```

## 2. Series (Tek Boyutlu Veri Yapısı)

`Series`, pandas'ın en temel veri yapısıdır. Tek bir sütundan oluşan, **indeksli** bir dizidir.

Aşağıda, veri setimizdeki `unit_price` (birim fiyat) sütununu tek başına bir Series olarak ele alacağız (veriyi az sonra tam olarak okuyacağız; burada kavramı göstermek için birkaç örnek değerle basit bir Series oluşturuyoruz).

```python
# Örnek birim fiyatlardan basit bir Series oluşturma
fiyatlar = pd.Series([7.46, 1.85, 7.38, 5.50, 8.66], name='unit_price')
print(fiyatlar)
print("\nVeri tipi:", type(fiyatlar))
print("Ortalama fiyat:", fiyatlar.mean())
```

## 3. DataFrame ve Veri Okuma/Yazma

`DataFrame`, satır ve sütunlardan oluşan **tablo yapısındaki** veri yapısıdır. Gerçek projelerde DataFrame genellikle bir dictionary'den değil, **dosyadan okunarak** elde edilir.

- `pd.read_json()` → JSON dosyasını okur
- `pd.read_excel()` → Excel dosyasını okur (sheet adı `sheet_name` ile belirtilebilir)
- `df.to_csv()` → DataFrame'i CSV olarak kaydeder

```python
# JSON dosyasından okuma
df_json = pd.read_json('grocery_chain_data.json')
print("JSON'dan okunan veri boyutu:", df_json.shape)
df_json.head()
```

```python
# Excel dosyasından okuma
df = pd.read_excel('grocery_chain_data.xlsx', sheet_name='Grocery_Data')
print("Excel'den okunan veri boyutu:", df.shape)
df.head()
```

```python
# DataFrame'i CSV olarak kaydetme
df.to_csv('grocery_chain_data.csv', index=False)
print("CSV olarak kaydedildi: grocery_chain_data.csv")
```

## 5. Veri Seçme ve Filtreleme

- **Sütun seçme**: `df['sutun']` veya `df[['sutun1','sutun2']]`
- **`.loc[]`**: Etiket bazlı seçim
- **`.iloc[]`**: Konum bazlı seçim
- **Koşullu filtreleme**: `df[df['sutun'] == deger]`

```python
# Sadece belirli sütunları seçme
df[['store_name', 'product_name', 'final_amount']].head()
```

```python
# .loc ile ilk satırın belirli sütunlarını seçme
df.loc[0, ['store_name', 'product_name', 'final_amount']]
```

```python
# Koşullu filtreleme: 'Dairy' (Süt Ürünleri) reyonundaki alışverişler
df[df['aisle'] == 'Dairy'].head()
```

```python
# Birden fazla koşul: final_amount 50 TL'den büyük VE reyon 'Meat & Seafood' olanlar
df[(df['final_amount'] > 50) & (df['aisle'] == 'Meat & Seafood')].head()
```

## 6. Veri Düzenleme

- **Yeni sütun ekleme**: Mevcut sütunlardan türetme (örn. indirim yüzdesi)
- **`.apply()`**: Her satıra özel bir fonksiyon uygular
- **Sütun silme / yeniden adlandırma**

```python
# İndirim yüzdesini hesaplayan yeni bir sütun ekleme
df['indirim_yuzdesi'] = (df['discount_amount'] / df['total_amount'] * 100).round(1)
df[['product_name', 'total_amount', 'discount_amount', 'indirim_yuzdesi']].head()
```

```python
# apply() ile kategorik bir sütun türetme: yüksek/düşük harcama etiketi
df['harcama_kategorisi'] = df['final_amount'].apply(lambda x: 'Yüksek' if x > 50 else 'Düşük')
df[['final_amount', 'harcama_kategorisi']].head()
```

```python
# Sütun ismini değiştirme (Türkçeleştirme örneği)
df_tr = df.rename(columns={'store_name': 'magaza_adi', 'product_name': 'urun_adi'})
df_tr[['magaza_adi', 'urun_adi']].head()
```

## 7. Sıralama ve Gruplama

- **Sıralama**: `df.sort_values('sutun', ascending=False)`
- **Gruplama (`groupby`)**: Kategorik bir sütuna göre grupları oluşturup toplama işlemi yapar
- **`.agg()`**: Aynı anda birden fazla istatistik hesaplar

```python
# En yüksek harcama tutarına göre sıralama
df.sort_values('final_amount', ascending=False)[['store_name', 'product_name', 'final_amount']].head()
```

```python
# Mağazaya göre gruplayıp ortalama harcamayı hesaplama
df.groupby('store_name')['final_amount'].mean().sort_values(ascending=False)
```

```python
# Reyona göre gruplayıp birden fazla istatistik (toplam ciro, işlem sayısı, ortalama)
df.groupby('aisle').agg(
    toplam_ciro=('final_amount', 'sum'),
    islem_sayisi=('final_amount', 'count'),
    ortalama_harcama=('final_amount', 'mean')
).sort_values('toplam_ciro', ascending=False)
```

## 8. Birleştirme İşlemleri (Concat & Merge)

- **`pd.concat([df1, df2])`**: DataFrame'leri alt alta birleştirir.
- **`pd.merge(df1, df2, on='ortak_sutun', how='...')`**: SQL'deki JOIN işlemine benzer.

Aşağıda, mağazalara ait ekstra bilgi (örn. bölge) içeren küçük bir tablo oluşturup gerçek veriyle birleştiriyoruz.

```python
# concat örneği: iki alt kümeyi alt alta birleştirme
df_dairy = df[df['aisle'] == 'Dairy'].head(3)
df_bakery = df[df['aisle'] == 'Bakery'].head(3)

df_birlesik = pd.concat([df_dairy, df_bakery])
df_birlesik[['store_name', 'aisle', 'product_name']]
```

```python
# merge örneği: mağazalara bölge bilgisi ekleme
magaza_bolgeleri = pd.DataFrame({
    'store_name': df['store_name'].dropna().unique(),
})
# Örnek bölge ataması (gerçek veride bulunmadığı için gösterim amaçlı ekliyoruz)
magaza_bolgeleri['bolge'] = ['Kuzey', 'Güney', 'Doğu', 'Batı', 'Kuzey', 'Güney', 'Doğu', 'Batı', 'Merkez'][:len(magaza_bolgeleri)]

df_bolgeli = pd.merge(df, magaza_bolgeleri, on='store_name', how='left')
df_bolgeli[['store_name', 'bolge', 'final_amount']].head()
```

## 9. Uygulamalı Örnek: Mağaza Bazlı Satış Analizi

Öğrenilen tüm kavramları birleştirip gerçek veri üzerinde küçük bir analiz yapalım: her mağazanın toplam cirosunu, işlem sayısını ve en çok satılan ürününü bulalım.

```python
# Mağaza bazında toplam ciro ve işlem sayısı
magaza_ozet = df.groupby('store_name').agg(
    toplam_ciro=('final_amount', 'sum'),
    islem_sayisi=('final_amount', 'count')
).sort_values('toplam_ciro', ascending=False)

print("Mağaza bazında özet:")
print(magaza_ozet)

en_iyi_magaza = magaza_ozet['toplam_ciro'].idxmax()
print(f"\nEn yüksek ciroya sahip mağaza: {en_iyi_magaza}")
```

```python
# En çok satılan ürün (toplam adet bazında)
en_cok_satilan = df.groupby('product_name')['quantity'].sum().sort_values(ascending=False)
print("En çok satılan ilk 5 ürün (toplam adet):")
print(en_cok_satilan.head())
```
