```python
import pandas as pd

# 5 yılın dosyasını ayrı ayrı okuyup bir sözlükte tutuyoruz
yillar = [2015, 2016, 2017, 2018, 2019]
ham_veriler = {yil: pd.read_csv(f"{yil}.csv") for yil in yillar}

for yil, df in ham_veriler.items():
    print(f"{yil}: {df.shape[0]} satır, {df.shape[1]} sütun")
```

```python
# SORUNU GÖRELİM: her yılın sütun isimleri birbirinden farklı!
for yil, df in ham_veriler.items():
    print(f"--- {yil} ---")
    print(list(df.columns), "\n")
```

```python
# Ortak şema: her yıldaki farklı ismi tek bir standart isme eşliyoruz
esleme = {
    2015: {"Country": "Ulke", "Happiness Rank": "Sira", "Happiness Score": "Skor",
           "Economy (GDP per Capita)": "GSYH", "Family": "Sosyal_Destek",
           "Health (Life Expectancy)": "Saglik", "Freedom": "Ozgurluk",
           "Trust (Government Corruption)": "Yolsuzluk_Algisi", "Generosity": "Comertlik"},
    2016: {"Country": "Ulke", "Happiness Rank": "Sira", "Happiness Score": "Skor",
           "Economy (GDP per Capita)": "GSYH", "Family": "Sosyal_Destek",
           "Health (Life Expectancy)": "Saglik", "Freedom": "Ozgurluk",
           "Trust (Government Corruption)": "Yolsuzluk_Algisi", "Generosity": "Comertlik"},
    2017: {"Country": "Ulke", "Happiness.Rank": "Sira", "Happiness.Score": "Skor",
           "Economy..GDP.per.Capita.": "GSYH", "Family": "Sosyal_Destek",
           "Health..Life.Expectancy.": "Saglik", "Freedom": "Ozgurluk",
           "Trust..Government.Corruption.": "Yolsuzluk_Algisi", "Generosity": "Comertlik"},
    2018: {"Country or region": "Ulke", "Overall rank": "Sira", "Score": "Skor",
           "GDP per capita": "GSYH", "Social support": "Sosyal_Destek",
           "Healthy life expectancy": "Saglik", "Freedom to make life choices": "Ozgurluk",
           "Perceptions of corruption": "Yolsuzluk_Algisi", "Generosity": "Comertlik"},
    2019: {"Country or region": "Ulke", "Overall rank": "Sira", "Score": "Skor",
           "GDP per capita": "GSYH", "Social support": "Sosyal_Destek",
           "Healthy life expectancy": "Saglik", "Freedom to make life choices": "Ozgurluk",
           "Perceptions of corruption": "Yolsuzluk_Algisi", "Generosity": "Comertlik"},
}

# Tüm yıllarda ortak olan sütunlar (bazı sütunlar her yılda olmadığı için sadece bunları tutacağız)
ortak_sutunlar = ["Ulke", "Sira", "Skor", "GSYH", "Sosyal_Destek",
                  "Saglik", "Ozgurluk", "Comertlik", "Yolsuzluk_Algisi"]

temiz_parcalar = []
for yil in yillar:
    df = ham_veriler[yil].rename(columns=esleme[yil])   # sütun isimlerini standartlaştır
    df["Yil"] = yil                                      # hangi yıla ait olduğunu ekle
    temiz_parcalar.append(df[["Yil"] + ortak_sutunlar])  # sadece ortak sütunları al

# concat ile 5 yılı tek DataFrame'de birleştiriyoruz
veri = pd.concat(temiz_parcalar, ignore_index=True)
print("Birleşik veri boyutu:", veri.shape)
veri.head()
```

```python
# Eksik değer var mı? Sütun bazında sayalım
veri.isnull().sum()
```

```python
# 2018'de Birleşik Arap Emirlikleri'nin yolsuzluk algısı değeri boş.
# Silmek yerine, o YILIN medyanı ile dolduruyoruz (derste öğrendiğimiz strateji)
yil_2018_medyan = veri.loc[veri["Yil"] == 2018, "Yolsuzluk_Algisi"].median()
print("2018 yılı medyanı:", yil_2018_medyan)

veri["Yolsuzluk_Algisi"] = veri["Yolsuzluk_Algisi"].fillna(yil_2018_medyan)
print("Doldurma sonrası eksik değer sayısı:", veri["Yolsuzluk_Algisi"].isnull().sum())
```

```python
print("Tamamen aynı olan satır sayısı        :", veri.duplicated().sum())
print("Aynı (Ülke, Yıl) ikilisine sahip satır:", veri.duplicated(subset=["Ulke", "Yil"]).sum())
# İkisi de 0 -> yinelenen kayıt yok. Olsaydı veri.drop_duplicates() ile temizlerdik.
```

```python
# Sira sütunu sıralama bilgisi olduğu için tam sayı (int) olmalı
veri["Sira"] = veri["Sira"].astype(int)
veri["Yil"] = veri["Yil"].astype(int)

veri.dtypes
```

```python
# Her ülkenin kaç kez geçtiğini sayalım; 5'ten az olanları listeleyelim
ulke_sayimi = veri["Ulke"].value_counts()
supheli = ulke_sayimi[ulke_sayimi < 5]
print("5'ten az görünen ülke sayısı:", len(supheli))
supheli
```

```python
# Listeye bakınca aynı ülkenin farklı yazımlarını görüyoruz:
# "Trinidad and Tobago" (3) + "Trinidad & Tobago" (2) = aslında 5!
# Bunları tek isimde birleştiriyoruz (replace ile)
isim_duzeltme = {
    "Trinidad & Tobago":        "Trinidad and Tobago",
    "Northern Cyprus":          "North Cyprus",
    "Taiwan Province of China": "Taiwan",
    "Hong Kong S.A.R., China":  "Hong Kong",
    "North Macedonia":          "Macedonia",
    "Somaliland Region":        "Somaliland region",
}
veri["Ulke"] = veri["Ulke"].replace(isim_duzeltme)

# Kontrol: düzeltme sonrası bu ülkeler artık 5'er kez görünüyor mu?
kontrol_listesi = ["Trinidad and Tobago", "North Cyprus", "Taiwan", "Hong Kong", "Macedonia"]
veri[veri["Ulke"].isin(kontrol_listesi)]["Ulke"].value_counts()
```

```python
# Yıl ve sıraya göre sıralayıp indeksi sıfırlıyoruz
veri = veri.sort_values(["Yil", "Sira"]).reset_index(drop=True)
veri.head()
```

```python
veri.to_csv("mutluluk_2015_2019_temiz.csv", index=False)   # index=False: indeks sütununu dosyaya yazma
print("Temiz veri kaydedildi:", veri.shape[0], "satır,", veri.shape[1], "sütun")
```

```python
def mutluluk_verisi_temizle(dosya_yollari: dict, esleme: dict, isim_duzeltme: dict) -> pd.DataFrame:
    """5 yıllık mutluluk raporu dosyalarını okuyup tek temiz DataFrame döndürür.

    Adımlar: sütun standartlaştırma -> birleştirme -> eksik doldurma ->
             duplicate kontrolü -> tip düzeltme -> isim tutarlılığı -> indeks düzenleme
    """
    ortak = ["Ulke", "Sira", "Skor", "GSYH", "Sosyal_Destek",
             "Saglik", "Ozgurluk", "Comertlik", "Yolsuzluk_Algisi"]

    parcalar = []
    for yil, yol in dosya_yollari.items():
        df = pd.read_csv(yol).rename(columns=esleme[yil])
        df["Yil"] = yil
        parcalar.append(df[["Yil"] + ortak])

    v = pd.concat(parcalar, ignore_index=True)

    # eksik değerleri kendi yılının medyanıyla doldur (groupby + transform)
    v["Yolsuzluk_Algisi"] = v.groupby("Yil")["Yolsuzluk_Algisi"].transform(lambda s: s.fillna(s.median()))

    v = v.drop_duplicates(subset=["Ulke", "Yil"])          # olası duplicate'leri at
    v["Sira"] = v["Sira"].astype(int)                       # tip düzeltme
    v["Ulke"] = v["Ulke"].replace(isim_duzeltme)            # isim tutarlılığı
    return v.sort_values(["Yil", "Sira"]).reset_index(drop=True)


# Fonksiyonu test edelim
dosyalar = {yil: f"{yil}.csv" for yil in yillar}
temiz = mutluluk_verisi_temizle(dosyalar, esleme, isim_duzeltme)
print("Fonksiyonla temizlenen veri:", temiz.shape)
print("Eksik değer:", temiz.isnull().sum().sum())
temiz.head()
```
